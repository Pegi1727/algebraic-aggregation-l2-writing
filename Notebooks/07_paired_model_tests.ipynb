# 07 — Paired model tests

**Purpose:** Wilcoxon signed-rank tests on supplied paired bootstrap MAE columns with Holm correction.

Self-contained, offline-runnable notebook invoking `l2_reproducibility_toolkit/07_paired_model_tests.py` via subprocess without modifying the original script. Generated artifacts are saved in `./outputs/notebooks/`.

```python
# Environment & deterministic settings
import sys, os, subprocess, json
from pathlib import Path
import pandas as pd
import numpy as np
REPO_ROOT = Path.cwd().resolve()
if REPO_ROOT.name == 'notebooks': REPO_ROOT = REPO_ROOT.parent
TOOLKIT = REPO_ROOT / 'l2_reproducibility_toolkit'
DATA_DIR = REPO_ROOT
OUT_DIR = Path.cwd() / 'outputs/notebooks'
OUT_DIR.mkdir(parents=True, exist_ok=True)
np.random.seed(42)
pd.set_option('display.max_columns', 50)
pd.set_option('display.precision', 4)
print('repo root:', REPO_ROOT, '| toolkit:', TOOLKIT, '| output:', OUT_DIR)
print('numpy', np.__version__, '| pandas', pd.__version__)
```

```python
# Load and inspect the relevant input file
INPUT = DATA_DIR / 'bootstrap_distributions_2000.csv'
df = pd.read_json(INPUT) if INPUT.suffix.lower()=='.json' else pd.read_csv(INPUT)
print(f"{INPUT.name}: shape={df.shape}")
df.head(10)
```

**Dependency note:** this script requires `scipy` in addition to pandas/numpy. If unavailable, install locally with `pip install scipy`; no network is used by the notebook itself.

```python
# Execute corresponding toolkit CLI (original script remains unchanged)
cmd = [sys.executable, str(TOOLKIT / '07_paired_model_tests.py'), '--input', str(INPUT), '--output', str(OUT_DIR / 'paired_model_tests.csv'), '--models', 'Additive_MAE,Godel_Min_MAE,Lukasiewicz_4D_MAE,Composite_Min_Add_MAE']
print('Running:', ' '.join(cmd))
res = subprocess.run(cmd, capture_output=True, text=True, cwd=str(TOOLKIT))
print(res.stdout)
if res.returncode:
    print('STDERR:', res.stderr)
    raise RuntimeError(f"Script failed with exit code {res.returncode}")
```

```python
# Display key result table
output_path = OUT_DIR / 'paired_model_tests.csv'
result = pd.read_csv(output_path) if output_path.exists() else pd.DataFrame()
print('Output:', output_path, '| shape:', result.shape)
result.head(15)
```
