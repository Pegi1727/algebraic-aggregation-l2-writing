# 10 — Publication figures

**Purpose:** Create an exploratory boxplot of numeric columns in the model comparison table.

Self-contained, offline-runnable notebook invoking `l2_reproducibility_toolkit/10_publication_figures.py` via subprocess without modifying the original script. Generated artifacts are saved in `./outputs/notebooks/`.

```python
# Environment & deterministic settings
import sys, os, subprocess, json
from pathlib import Path
import pandas as pd
import numpy as np
REPO_ROOT = Path.cwd().resolve()
if REPO_ROOT.name == 'notebooks': REPO_ROOT = REPO_ROOT.parent
TOOLKIT = REPO_ROOT / 'l2_reproducibility_toolkit'
DATA_DIR = REPO_ROOT
OUT_DIR = Path.cwd() / 'outputs/notebooks'
OUT_DIR.mkdir(parents=True, exist_ok=True)
np.random.seed(42)
pd.set_option('display.max_columns', 50)
pd.set_option('display.precision', 4)
print('repo root:', REPO_ROOT, '| toolkit:', TOOLKIT, '| output:', OUT_DIR)
print('numpy', np.__version__, '| pandas', pd.__version__)
```

```python
# Load and inspect the relevant input file
INPUT = DATA_DIR / 'final_model_comparison_APA.csv'
df = pd.read_json(INPUT) if INPUT.suffix.lower()=='.json' else pd.read_csv(INPUT)
print(f"{INPUT.name}: shape={df.shape}")
df.head(10)
```

**Output note:** the script creates a PNG at `outputs/notebooks/10_publication_figures.png`; the following cell displays it inline.

```python
# Execute corresponding toolkit CLI (original script remains unchanged)
cmd = [sys.executable, str(TOOLKIT / '10_publication_figures.py'), '--input', str(INPUT), '--output', str(OUT_DIR / '10_publication_figures.png')]
print('Running:', ' '.join(cmd))
res = subprocess.run(cmd, capture_output=True, text=True, cwd=str(TOOLKIT))
print(res.stdout)
if res.returncode:
    print('STDERR:', res.stderr)
    raise RuntimeError(f"Script failed with exit code {res.returncode}")
```

```python
from IPython.display import Image, display
figure_path = OUT_DIR / '10_publication_figures.png'
if figure_path.exists(): display(Image(filename=str(figure_path), width=800))
else: print('Figure missing; inspect the script output above.')
```
